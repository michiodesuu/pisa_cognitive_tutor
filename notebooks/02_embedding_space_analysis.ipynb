# Embedding Space Analysis

Analyze BGE-M3 embedding quality for Thai/Chinese/English PISA questions.

```python
import sys; sys.path.insert(0, '..')
from pathlib import Path
from src.retrieval.embedder import BGEEmbedder
import numpy as np, matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity

embedder = BGEEmbedder(Path('../configs/model_configs.yaml'))

# Test multilingual queries
queries = [
    'photosynthesis light energy conversion',
    'การสังเคราะห์ด้วยแสง พลังงานแสงอาทิตย์',  # Thai
    '光合作用 光能转化',                          # Chinese
    'greenhouse effect carbon dioxide warming',
    'ผลกระทบเรือนกระจก คาร์บอนไดออกไซด์',
    'evolution natural selection adaptation',
    'วิวัฒนาการ การคัดเลือกโดยธรรมชาติ',
]

dense_vecs = embedder.encode_dense(queries)
print(f'Dense vector shape per query: {len(dense_vecs[0])}')

# Check cross-lingual similarity
sim = cosine_similarity([dense_vecs[0]], [dense_vecs[1]])[0][0]
print(f'EN vs TH photosynthesis cosine similarity: {sim:.4f}  (>0.7 = good)')

# 2D projection
pca = PCA(n_components=2)
coords = pca.fit_transform(dense_vecs)
colors = ['blue','blue','blue','green','green','red','red']
fig, ax = plt.subplots(figsize=(8,5))
for (x,y), label, c in zip(coords, queries, colors):
    ax.scatter(x, y, c=c, s=80)
    ax.annotate(label[:30], (x,y), fontsize=7, ha='left')
ax.set_title('BGE-M3 Embedding Space (PCA 2D)')
plt.tight_layout(); plt.show()
```
